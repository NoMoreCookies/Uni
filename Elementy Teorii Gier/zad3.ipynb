```python
import numpy as np
import pandas as pd
from scipy.optimize import linprog
```

```python
np_matrix = np.random.randint(0, 100, size = (6,6))
diag = [2,8,2,2,5,5]
```

```python
for n, integer in enumerate(diag):
    np_matrix[n, n] = integer
```

```python
pd_matrix = pd.DataFrame(np_matrix)
```

```python
pd_matrix
```

```python
n = 12

mask = np.zeros(pd_matrix.size, dtype=bool)

indices = np.random.choice(pd_matrix.size, n, replace=False)

mask[indices] = True

pd_matrix = pd_matrix.mask(mask.reshape(pd_matrix.shape), -pd_matrix)
```

```python
pd_matrix
```

Problem programowania liniowego

$$
\begin{array}{ll}
\text{Maksymalizuj:} & v \\
\text{przy ograniczeniach:} & \sum_{i=1}^{m} a_{i,j} \cdot x_i \ge v, \quad j=1, 2, \dots, n \\
& \sum_{i=1}^{m} x_i = 1 \\
& x_i \ge 0, \quad i=1, 2, \dots, m
\end{array}
$$

```python
#1. Dane i stałe

A = pd_matrix.to_numpy()
m, n = A.shape

k = abs(A.min()) + 1  # bo linrog na dodatnich macierzach

A_plus = A + k
```

```python

# 2. Budujemy układ: -A.T * x + v <= 0
# Łączymy macierz z kolumną jedynek (dla v)
lewa_strona = np.column_stack([-A_plus.T, np.ones(n)])
prawa_strona = np.zeros(n)
```

```python

# 3. Suma prawdopodobieństw (x) musi być 1
c_row = np.append(np.ones(m), 0) 
c_eq = [1]
```

```python

# 4. Cel: Max v (czyli Min -v)
cel = np.append(np.zeros(m), -1)
```

```python

# 5. Granice: x to [0, 1], v to cokolwiek
granice = [(0, 1)] * m + [(None, None)]
```

```python

# 6. Rozwiązanie
res = linprog(cel, A_ub=lewa_strona, b_ub=prawa_strona, 
              A_eq=[c_row], b_eq=c_eq, bounds=granice)
```

```python

if res.success:
    v = res.x[-1] - k
    strategia = res.x[:m].round(3)
    print(f"Wartość gry: {v} | Strategia: {strategia}")
```
